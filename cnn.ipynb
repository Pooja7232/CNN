```python
#Importing Libraries
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import numpy as np
import pandas as pd
```

```python
#Importing the dataset
dataset_path=r"C:\Users\POOJASRIMENDU\Downloads\ImageDatasets\PetImages"
```

```python
#Image preprocessing
datagen=ImageDataGenerator(
    #Normalize pixels
    rescale=1./255,
    #Data augmentation
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    #Split dataset
    validation_split=0.2 #Validation split used to create train and validation data
)
```

```python
#Load Training Data
#Training data
train_data=datagen.flow_from_directory(
    dataset_path,
    target_size=(128,128),
    batch_size=32,
    class_mode='binary',
    subset='training'
)
```

```python
#LOad Validation Data
val_data=datagen.flow_from_directory(
    dataset_path,
    target_size=(128,128),
    batch_size=32,
    class_mode='binary',
    subset='validation'
)
```

```python
print(train_data.class_indices)
```

```python
#Building CNN Model
model=keras.Sequential()
```

```python
#First Convolutional Layer
model.add(
    keras.layers.Conv2D(
    filters=32,
    kernel_size=(3,3),
    activation='relu',
    input_shape=(128,128,3)
    )
)
```

```python
#First Pooling Layer
model.add(
    keras.layers.MaxPooling2D(
        pool_size=(2,2)
    )
)
```

```python
#Second Convlolution Layer
model.add(
    keras.layers.Conv2D(
        filters=32,
        kernel_size=(3,3),
        activation='relu'
    )
)
```

```python
#Second Pooling Layer
model.add(
    keras.layers.MaxPooling2D(
        pool_size=(2,2)
    )
)
```

```python
#Third Convlolution Layer
model.add(
    keras.layers.Conv2D(
        filters=32,
        kernel_size=(3,3),
        activation='relu'
    )
)
```

```python
#Third Pooling Layer
model.add(
    keras.layers.MaxPooling2D(
        pool_size=(2,2)
    )
)
```

```python
#Flatten Layer
model.add(keras.layers.Flatten())
```

```python
#Dense Layer
model.add(
    keras.layers.Dense(
        units=128,
        activation='relu'
    )
)
```

```python
#Output layer
model.add(
    keras.layers.Dense(
        units=1,
        activation='sigmoid'
    )
)
```

```python
model.summary()
```

```python
print(tf.config.list_physical_devices())
```

```python
#Compile CNN
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
#Train
history=model.fit(
    train_data,
    epochs=10,
    validation_data=val_data
)
```

```python
#Evaluate the model
loss,accuracy=model.evaluate(val_data)
print("Accuracy:",accuracy)
```
